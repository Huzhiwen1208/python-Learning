# 一、课程导入：为什么需要面向对象？

在没有学习类之前，我们通常用变量、列表、字典和函数来组织程序。

例如，我们想描述一只狗：

- 名字：旺财
- 年龄：3
- 行为：会叫

可以用变量和函数这样写。

````python
dog_name = "旺财"
dog_age = 3


def dog_bark(name: str) -> None:
    print(f"{name}：汪汪汪！")


print(dog_name)
print(dog_age)
dog_bark(dog_name)

dog_dict = {
    "dog_name": "旺财",
    "dog_age": 3,
    "dog_bark": dog_bark
}
````

这段代码可以运行，但如果我们有很多只狗，就会变得很混乱。

```python
dog1_name = "旺财"
dog1_age = 3

dog2_name = "小黑"
dog2_age = 2

dog3_name = "豆豆"
dog3_age = 5
```

问题是：**数据和行为是分散的**。

面向对象的思想是：

> 把相关的数据和行为放在一起，用一个“对象”表示现实世界中的一个事物。

例如：

- 一只狗是一个对象
- 一个学生是一个对象
- 一个商品是一个对象
- 一个银行账户是一个对象
- 一个订单是一个对象

# 二、类和对象的基本概念

## 1. 什么是类？

类，英文是 `class`，可以理解为**模板**或**图纸**。

例如：

- `Dog` 类：描述所有狗应该有哪些属性和行为
- `Student` 类：描述所有学生应该有哪些属性和行为
- `Product` 类：描述所有商品应该有哪些属性和行为

## 2. 什么是对象？

对象是根据类创建出来的一个具体实例。

例如：

- `Dog` 是类
- `旺财` 是一个具体的狗对象
- `小黑` 也是一个具体的狗对象

一句话总结：

> 类是模板，对象是根据模板创建出来的具体东西。

## 3. 第一个类：`Dog`

下面我们用 `class` 定义一个狗类。

````python
class Dog:
    def bark(self) -> None:
        print("汪汪汪！")


# 创建对象
wangcai = Dog()

# 调用对象的方法
wangcai.bark()
````

上面的代码中：

```python
class Dog:
```

表示定义一个类，类名叫 `Dog`。

```python
def bark(self):
```

表示定义一个方法。方法可以简单理解为：**写在类里面的函数**。

```python
wangcai = Dog()
```

表示根据 `Dog` 这个类创建一个对象，变量名叫 `wangcai`。

# 三、`__init__`、`self`、属性和方法

刚才的 `Dog` 类只能叫，但是还没有名字和年龄。

如果希望每只狗都有自己的名字和年龄，就需要使用 `__init__` 方法。

## 1. `__init__` 是什么？

`__init__` 是一个特殊方法，会在创建对象时自动执行。

它通常用于给对象设置初始属性。

````python
class Dog:
    def __init__(self, name: str, age: int) -> None:
        self.name = name
        self.age = age

    def bark(self) -> None:
        print(f"{self.name}：汪汪汪！")
    
    def sleep(self) -> None:
        # 直接睡一年, 爱说梦话
        self.age += 1
        self.bark()

wangcai = Dog("旺财", 3)
xiaohei = Dog("小黑", 2)

print(wangcai.name, wangcai.age)
print(xiaohei.name, xiaohei.age)

wangcai.bark()
xiaohei.bark()

wangcai.sleep()
````

## 2. `self` 是什么？

`self` 表示当前对象本身。

当我们写：

```python
self.name = name
self.age = age
```

意思是：

- 给当前对象保存一个 `name` 属性
- 给当前对象保存一个 `age` 属性

当我们创建两个对象时：

```python
wangcai = Dog("旺财", 3)
xiaohei = Dog("小黑", 2)
```

它们有各自独立的 `name` 和 `age`。

## 3. 属性和方法

在类中，常见的两个概念是：

| 概念 | 含义 | 示例 |
|---|---|---|
| 属性 | 对象拥有的数据 | 名字、年龄、分数、余额 |
| 方法 | 对象可以执行的行为 | 叫、学习、存钱、查询余额 |

例如在 `Dog` 类中：

```python
self.name
self.age
```

是属性。

```python
def bark(self):
```

是方法。

## 4. 方法可以修改对象状态

对象不仅可以保存数据，还可以通过方法修改自己的数据。

````python
class Student:
    def __init__(self, name: str, score: int) -> None:
        self.name = name
        self.score = score

    def add_score(self, points: int) -> None:
        self.score += points

    def show_info(self) -> None:
        print(f"学生：{self.name}，分数：{self.score}")


# 创建具体对象
student = Student("小明", 80)
# 调用行为
student.show_info()

# 行为
student.add_score(5)
# 行为
student.show_info()
````

这里的 `add_score()` 方法会修改对象自己的 `score` 属性。

这就是对象的一个重要特点：

> 对象可以保存自己的状态，也可以通过方法改变自己的状态。

# 五、类属性、实例属性和简单封装

## 1. 实例属性

实例属性是属于某一个具体对象的属性。

例如：

```python
self.name
self.score
```

每个对象都可以有不同的值。

````python
class Student:
    def __init__(self, name: str, score: int) -> None:
        self.name = name
        self.score = score


s1 = Student("小明", 90)
s2 = Student("小红", 85)

print(s1.name, s1.score)
print(s2.name, s2.score)
````

## 2. 类属性

类属性是属于整个类的属性，所有对象共享。

例如：所有学生都属于同一所学校。

````python
class Student:
    school = "Python 编程学院"  # 类属性

    def __init__(self, name: str, score: int) -> None:
        self.name = name      # 实例属性
        self.score = score    # 实例属性


s1 = Student("小明", 90)
s2 = Student("小红", 85)

print(s1.school)
print(s2.school)
print(Student.school)
````

## 3. 用类属性统计对象数量

类属性也常用于记录全局信息，例如创建了多少个学生对象。

````python
class Student:
    count = 0

    def __init__(self, name: str, score: int) -> None:
        self.name = name
        self.score = score
        Student.count += 1


s1 = Student("小明", 90)
s2 = Student("小红", 85)
s3 = Student("小刚", 78)

s1.score += 30

print("学生对象数量：", Student.count)
````

## 4. 简单封装

封装的意思是：

> 把数据和操作数据的方法放在一起，并尽量通过方法来访问和修改数据。

例如银行账户的余额，不应该随便被外部修改。

在 Python 中，以下划线开头的属性，例如 `_balance`，通常表示：这个属性是内部使用的，不建议外部直接修改。

````python
class BankAccount:
    def __init__(self, owner: str, balance: float) -> None:
        self.owner = owner
        self._balance = balance

    def deposit(self, amount: float) -> None:
        if amount <= 0:
            print("存款金额必须大于 0")
            return
        self._balance += amount

    def withdraw(self, amount: float) -> None:
        if amount <= 0:
            print("取款金额必须大于 0")
            return
        if amount > self._balance:
            print("余额不足")
            return
        self._balance -= amount

    def get_balance(self) -> float:
        return self._balance



account = BankAccount("小明", 1000)
account.deposit(500)
account.withdraw(200)
print(account.get_balance())
````

注意：Python 不会完全禁止你访问 `_balance`，但下划线是一种约定，告诉其他程序员：

> 这个属性不建议直接从外部修改，请通过方法来操作。

## 5. `__str__` 方法

当我们直接打印对象时，默认结果通常不够友好。

````python
class Student:
    def __init__(self, name: str, score: int) -> None:
        self.name = name
        self.score = score


student = Student("小明", 90)
print(student)
````

可以通过定义 `__str__` 方法，让对象打印时更加清晰。

````python
class Student:
    def __init__(self, name: str, score: int) -> None:
        self.name = name
        self.score = score

    def __str__(self) -> str:
        return f"Student(name={self.name}, score={self.score})"

student = Student("小明", 90)
print(student)
````

# 六、综合案例：学生成绩类

接下来我们用面向对象写一个简单的学生成绩管理示例。

需求：

1. 每个学生有姓名和成绩
2. 可以判断学生等级
3. 可以打印学生信息
4. 可以从多个学生中找出最高分学生

````python
class Student:
    def __init__(self, name: str, score: int) -> None:
        self.name = name
        self.score = score

    def get_level(self) -> str:
        if self.score >= 90:
            return "优秀"
        elif self.score >= 80:
            return "良好"
        elif self.score >= 60:
            return "及格"
        else:
            return "不及格"

    def show_info(self) -> None:
        print(f"{self.name}：{self.score} 分，等级：{self.get_level()}")


students = [
    Student("小明", 95),
    Student("小红", 87),
    Student("小刚", 72),
    Student("小丽", 58),
]

for student in students:
    student.show_info()
````

## 找出最高分学生

对象可以放进列表中，然后像处理普通列表一样处理对象列表。

````python
top_student = max(students, key=lambda student: student.score)

print("最高分学生：")
top_student.show_info()
````

## 进一步封装：班级类 `ClassRoom`

如果我们希望“班级”也成为一个对象，就可以定义 `ClassRoom` 类。

一个班级对象可以：

- 保存多个学生
- 添加学生
- 打印所有学生成绩
- 计算平均分
- 找出最高分学生

````python
class Student:
    def __init__(self, name: str, score: int) -> None:
        self.name = name
        self.score = score

    def get_level(self) -> str:
        if self.score >= 90:
            return "优秀"
        elif self.score >= 80:
            return "良好"
        elif self.score >= 60:
            return "及格"
        else:
            return "不及格"

    def __str__(self) -> str:
        return f"{self.name}：{self.score} 分，等级：{self.get_level()}"


class ClassRoom:
    def __init__(self, name: str) -> None:
        self.name = name
        self.students: list[Student] = []

    def add_student(self, student: Student) -> None:
        self.students.append(student)

    def show_all_students(self) -> None:
        print(f"班级：{self.name}")
        for student in self.students:
            print(student)

    def get_average_score(self) -> float:
        if len(self.students) == 0:
            return 0
        total = sum(student.score for student in self.students)
        return total / len(self.students)

    def get_top_student(self):
        if len(self.students) == 0:
            return None
        return max(self.students, key=lambda student: student.score)


classroom = ClassRoom("Python 基础班")
classroom.add_student(Student("小明", 95))
classroom.add_student(Student("小红", 87))
classroom.add_student(Student("小刚", 72))
classroom.add_student(Student("小丽", 58))

classroom.show_all_students()
print("平均分：", round(classroom.get_average_score(), 2))

student = classroom.get_top_student()
if student is not None:
    print("最高分学生：", student)
````

这个案例中有两个类：

| 类 | 作用 |
|---|---|
| `Student` | 表示一个学生 |
| `ClassRoom` | 表示一个班级，管理多个学生对象 |

这就是面向对象程序设计中很常见的方式：

> 用不同的类表示不同的事物，让对象之间相互协作。

## 课堂练习 2：参考答案

````python
class Rectangle:
    def __init__(self, width: float, height: float) -> None:
        self.width = width
        self.height = height

    def get_area(self) -> float:
        return self.width * self.height

    def get_perimeter(self) -> float:
        return 2 * (self.width + self.height)

    def resize(self, width: float, height: float) -> None:
        self.width = width
        self.height = height

    def __str__(self) -> str:
        return f"Rectangle(width={self.width}, height={self.height})"


rectangle = Rectangle(10, 5)
print(rectangle.get_area())
print(rectangle.get_perimeter())
print(rectangle)

rectangle.resize(20, 8)
print(rectangle.get_area())
print(rectangle)
````

````python
class Student:
    def __init__ (self,name: str,score: dict[str,int]) -> None:
        self.name = name
        self.score =score

    def __str__(self):
        return f"姓名：{self.name}，分数：{self.score}"

class ClassRoom:
    def __init__ (self,name: str,students: list[Student]) ->None:
        self.name = name
        self.students = students


        
    def add_student(self,student_name: str, student_score: dict[str,int]) -> None:
        new_student = Student(student_name, student_score)
        self.students.append(new_student)
    

    def show_all_students(self) -> None:
        print(self.students)

    def get_class_average_score(self) -> float:
        if len(self.students) == 0:
            return None
        total_all = 0  
        total_subject = 0 
        for stu in self.students:
        
            total_all += sum(stu.score.values())
            total_subject += len(stu.score)

        average = total_all / total_subject
        return average
    
    def get_top_student(self):
        if len(self.students) == 0:
            return None
        return max(self.students, key=lambda s: sum(s.score.values()))




classroom = ClassRoom("Python OOP 班",[])


classroom.add_student("小明", {"语文": 90, "数学": 95, "英语": 92})
classroom.add_student("小红", {"语文": 85, "数学": 88, "英语": 90})
classroom.add_student("小刚", {"语文": 70, "数学": 65, "英语": 72})

classroom.show_all_students()
print("全班平均分：", round(classroom.get_class_average_score(), 2))
print("最高分学生：", classroom.get_top_student())
````
