# 第二部分：读取文本文件

Python 使用 `open()` 函数打开文件。

基本格式：

```python
open("文件名", "模式", encoding="编码")
```

例如：

```python
open("note.txt", "r", encoding="utf-8")
```

这里：

- `"note.txt"` 表示文件名；
- `"r"` 表示读取模式；
- `encoding="utf-8"` 表示使用 UTF-8 编码，适合处理中英文文本。

## 1. 先创建一个演示文本文件

在 Notebook 中，我们可以先用代码创建一个文件，方便后面读取。

````python
with open("note.txt", "w", encoding="utf-8") as file:
    file.write("Hello Python\n")
    file.write("这是一个文本文件\n")
    file.write("今天学习文件读写\n")
````

运行上面的代码后，当前目录下会生成一个 `note.txt` 文件。

下面我们来读取它。

## 2. 使用 `open()` 和 `read()` 读取全部内容

最基础的读取方式是：

```python
file = open(...)
content = file.read()
file.close()
```

注意：

文件打开后，最后要关闭文件。  
如果忘记关闭，可能造成资源浪费。

````python
file = open("note.txt", "r", encoding="utf-8")
content = file.read()
file.close()

print(content)
````

这段代码可以正常运行，但不够推荐。

因为如果中间代码出错，`file.close()` 可能不会执行。

所以更推荐使用 `with open()`。

## 3. 推荐写法：`with open()`

`with open()` 的好处是：

> 文件使用完后，会自动关闭。

这是实际开发中更常见、更推荐的写法。

````python
with open("note.txt", "r", encoding="utf-8") as file:
    content = file.read()

print(content)
````

## 讲解重点

`with open()` 可以理解为：

> 打开文件，执行缩进代码块中的内容，执行结束后自动关闭文件。

以后只要进行文件读写，优先使用这种写法：

```python
with open("文件名", "模式", encoding="utf-8") as file:
    ...
```

# 第三部分：写入文本文件

除了读取文件，程序也可以向文件中写入内容。

写入文件使用模式 `"w"`。

```python
with open("文件名", "w", encoding="utf-8") as file:
    file.write("内容")
```

注意：

> `"w"` 模式会覆盖文件原来的内容。

## 1. 写入文件

下面创建一个新的文本文件 `message.txt`。

````python
with open("message.txt", "w", encoding="utf-8") as file:
    file.write("第一行：学习 Python 文件写入\n")
    file.write("第二行：使用 write 方法\n")

print("写入完成")
````

读取刚才写入的文件。

````python
with open("message.txt", "r", encoding="utf-8") as file:
    content = file.read()

print(content)
````

## 2. 覆盖写入

如果再次使用 `"w"` 模式写入同一个文件，原来的内容会被清空。

````python
with open("message.txt", "w", encoding="utf-8") as file:
    file.write("这是新的内容，原来的内容已经被覆盖。\n")

with open("message.txt", "r", encoding="utf-8") as file:
    content = file.read()

print(content)
````

## 3. 追加写入

如果不想覆盖原来的内容，而是想在文件末尾继续添加内容，可以使用 `"a"` 模式。

`a` 是 append 的意思，表示追加。

````python
with open("message.txt", "a", encoding="utf-8") as file:
    file.write("这是追加的新内容。\n")
    file.write("追加模式不会删除原来的内容。\n")

with open("message.txt", "r", encoding="utf-8") as file:
    content = file.read()

print(content)
````

## 常见文件打开模式

| 模式 | 含义 |
|---|---|
| `"r"` | 读取文件，文件必须存在 |
| `"w"` | 写入文件，如果文件存在则覆盖，不存在则创建 |
| `"a"` | 追加写入，如果文件不存在则创建 |
| `"r+"` | 读写模式，文件必须存在 |

初学阶段最常用的是：

```python
"r"
"w"
"a"
```

# 第四部分：按行读取文件

有时候我们不想一次性读取整个文件，而是希望一行一行处理。

例如，一个学生成绩文件：

```text
Tom,95
Alice,88
Jack,76
```

每一行都是一个学生的数据。

这种情况就适合按行读取。

## 1. 创建学生成绩文本文件

````python
with open("students.txt", "w", encoding="utf-8") as file:
    file.write("Tom,95\n")
    file.write("Alice,88\n")
    file.write("Jack,76\n")
    file.write("Lily,91\n")
````

## 2. 逐行读取

可以直接遍历文件对象。

````python
with open("students.txt", "r", encoding="utf-8") as file:
    for line in file:
        print(line)
````

你可能会发现输出中有多余的空行。

这是因为每一行本身已经带有换行符 `\n`，而 `print()` 默认又会换行。

所以我们通常使用 `strip()` 去掉首尾空白字符。

````python
with open("students.txt", "r", encoding="utf-8") as file:
    for line in file:
        print(line.strip())
````

## 3. 把文本解析成字典

现在我们想把文件内容：

```text
Tom,95
Alice,88
Jack,76
Lily,91
```

转换成 Python 字典：

```python
{
    "Tom": 95,
    "Alice": 88,
    "Jack": 76,
    "Lily": 91
}
```

可以使用：

```python
split(",")
```

````python
scores: dict[str, int] = {}

with open("students.txt", "r", encoding="utf-8") as file:
    for line in file:
        name, score = line.strip().split(",")
        scores[name] = int(score)

print(scores)
````

## 这段代码的逻辑

以这一行为例：

```text
Tom,95
```

执行：

```python
line.strip()
```

得到：

```python
"Tom,95"
```

执行：

```python
split(",")
```

得到：

```python
["Tom", "95"]
```

再通过：

```python
name, score = ...
```

拆成两个变量。

最后：

```python
scores[name] = int(score)
```

把字符串分数转换成整数，保存到字典中。

# 第五部分：异常处理

程序运行过程中，经常会遇到错误。

例如：

- 文件不存在；
- 输入内容不是数字；
- 字典中没有某个键；
- 列表下标越界；
- 除数为 0。

如果不处理这些错误，程序会直接崩溃。

异常处理的作用是：

> 当程序出现可预期错误时，不让程序直接崩溃，而是给出友好的提示或执行备用逻辑。

## 1. 文件不存在错误

下面这段代码会尝试读取一个不存在的文件。

````python
# 这段代码如果直接运行，会出现 FileNotFoundError
# 为了避免 Notebook 中断，这里先注释掉

# with open("not_exist.txt", "r", encoding="utf-8") as file:
#     content = file.read()
# 
# print(content)
````

如果运行上面的代码，会出现类似错误：

```text
FileNotFoundError: [Errno 2] No such file or directory: 'not_exist.txt'
```

意思是：文件不存在。

## 2. 使用 `try-except` 处理异常

基本语法：

```python
try:
    可能出错的代码
except 错误类型:
    出错后的处理逻辑
```

````python
try:
    with open("not_exist.txt", "r", encoding="utf-8") as file:
        content = file.read()
    print(content)
except FileNotFoundError:
    print("文件不存在，请检查文件名。")
````

现在，即使文件不存在，程序也不会崩溃，而是输出友好提示。

## 3. 处理输入错误

`input()` 得到的是字符串。

如果用户输入的不是数字，使用 `int()` 转换时会出错。

在 Notebook 中，为了方便演示，我们用变量模拟用户输入。

````python
user_input = "abc"

try:
    age = int(user_input)
    print("年龄是：", age)
except ValueError:
    print("输入内容不是合法数字。")
````

如果 `user_input = "18"`，代码就可以正常转换。

````python
user_input = "18"

try:
    age = int(user_input)
    print("年龄是：", age)
except ValueError:
    print("输入内容不是合法数字。")
````

## 4. 常见异常类型

| 异常类型 | 常见原因 |
|---|---|
| `FileNotFoundError` | 文件不存在 |
| `ValueError` | 值不合法，例如 `int("abc")` |
| `ZeroDivisionError` | 除数为 0 |
| `KeyError` | 字典中没有指定键 |
| `IndexError` | 列表下标越界 |
| `TypeError` | 类型不支持某个操作 |

初学时不需要记住所有异常，重点是理解：

> 程序出错不一定只能崩溃，我们可以提前设计处理方案。

## 5. 同时处理多个异常

可以写多个 `except`。

````python
data = {"Tom": "95", "Alice": "88"}

try:
    score = int(data["Jack"])
    print(score)
except KeyError:
    print("没有找到这个学生。")
except ValueError:
    print("成绩不是合法数字。")
````

# 第六部分：JSON 文件读写

前面我们使用普通文本保存数据。

普通文本适合简单内容，例如：

```text
Tom,95
Alice,88
```

但是如果数据更复杂，比如：

```python
books = [
    {"title": "Python基础", "author": "张三", "is_borrowed": False},
    {"title": "数据分析入门", "author": "李四", "is_borrowed": True}
]
```

用普通文本保存会比较麻烦。

这时候更适合使用 JSON。

JSON 是一种常见的数据交换格式，特点是：

- 结构清晰；
- 很适合保存列表和字典；
- 很多编程语言都支持；
- 常用于配置文件、接口数据、项目数据保存。

## 1. Python 数据与 JSON

Python 中常见数据可以保存成 JSON：

| Python | JSON |
|---|---|
| `dict` | object |
| `list` | array |
| `str` | string |
| `int` / `float` | number |
| `True` / `False` | true / false |
| `None` | null |

Python 使用内置模块 `json` 处理 JSON 文件。

```python
import json
```

## 2. 写入 JSON 文件

使用：

```python
json.dump(数据, 文件对象, ensure_ascii=False, indent=4)
```

其中：

- `ensure_ascii=False`：保证中文正常显示；
- `indent=4`：让 JSON 文件更容易阅读。

````python
import json

books = [
    {"title": "Python基础", "author": "张三", "is_borrowed": False},
    {"title": "数据分析入门", "author": "李四", "is_borrowed": True},
    {"title": "机器学习实战", "author": "王五", "is_borrowed": False}
]

with open("books.json", "w", encoding="utf-8") as file:
    json.dump(books, file, ensure_ascii=False, indent=4)

print("JSON 文件写入完成")
````

## 3. 读取 JSON 文件

使用：

```python
json.load(文件对象)
```

会把 JSON 文件内容读取成 Python 数据。

````python
import json

with open("books.json", "r", encoding="utf-8") as file:
    loaded_books = json.load(file)

print(loaded_books)
print(type(loaded_books))
print(type(loaded_books[0]))
````

可以看到，读取回来之后：

- 整体是 `list`；
- 每一本书是 `dict`。

这说明 JSON 非常适合保存 Python 中的列表和字典。

# 第七部分：综合案例，图书管理系统支持保存和加载

## 1. `Book` 类设计

一本书包含：

- 书名 `title`
- 作者 `author`
- 是否借出 `is_borrowed`

另外，因为 JSON 不能直接保存对象，所以我们需要把 `Book` 对象转换成字典。

因此设计一个方法：

```python
to_dict()
```

用于把对象转换成字典。

再设计一个方法：

```python
from_dict()
```

用于把字典转换回 `Book` 对象。

````python
class Book:
    def __init__(self, title: str, author: str, is_borrowed: bool = False):
        self.title = title
        self.author = author
        self.is_borrowed = is_borrowed

    def borrow(self) -> bool:
        if self.is_borrowed:
            return False

        self.is_borrowed = True
        return True

    def return_book(self) -> bool:
        if not self.is_borrowed:
            return False

        self.is_borrowed = False
        return True

    def to_dict(self) -> dict:
        return {
            "title": self.title,
            "author": self.author,
            "is_borrowed": self.is_borrowed
        }

    @staticmethod
    def from_dict(data: dict) -> "Book":
        return Book(
            title=data["title"],
            author=data["author"],
            is_borrowed=data["is_borrowed"]
        )

    def __str__(self) -> str:
        status = "已借出" if self.is_borrowed else "可借阅"
        return f"《{self.title}》 - {self.author} - {status}"
````

## 2. 测试 `Book` 类

````python
book = Book("Python基础", "张三")

print(book)
print(book.to_dict())

book.borrow()
print(book)

book.return_book()
print(book)
````

## 3. `Library` 类设计

`Library` 类负责管理多本书。

5. 保存到 JSON 文件；
6. 从 JSON 文件加载。

````python
import json


class Library:
    def __init__(self):
        self.books: list[Book] = []

    def add_book(self, book: Book) -> None:
        self.books.append(book)

    def show_books(self) -> None:
        if len(self.books) == 0:
            print("图书馆暂无图书。")
            return

        for book in self.books:
            print(book)

    def find_book(self, title: str) -> Book | None:
        for book in self.books:
            if book.title == title:
                return book
        return None

    def borrow_book(self, title: str) -> bool:
        book = self.find_book(title)

        if book is None:
            print("没有找到这本书。")
            return False

        if book.borrow():
            print(f"借阅成功：{title}")
            return True
        else:
            print(f"借阅失败：{title} 已被借出")
            return False

    def return_book(self, title: str) -> bool:
        book = self.find_book(title)

        if book is None:
            print("没有找到这本书。")
            return False

        if book.return_book():
            print(f"归还成功：{title}")
            return True
        else:
            print(f"归还失败：{title} 当前并未借出")
            return False

    def save_to_file(self, filename: str) -> None:
        data = [book.to_dict() for book in self.books]

        with open(filename, "w", encoding="utf-8") as file:
            json.dump(data, file, ensure_ascii=False, indent=4)

        print(f"图书数据已保存到 {filename}")

    def load_from_file(self, filename: str) -> None:
        try:
            with open(filename, "r", encoding="utf-8") as file:
                data = json.load(file)

            self.books = [Book.from_dict(item) for item in data]
            print(f"已从 {filename} 加载图书数据")

        except FileNotFoundError:
            print("文件不存在，将使用空图书馆。")
            self.books = []
````

## 4. 测试完整图书管理系统

下面演示：

1. 创建图书馆；
2. 添加图书；
3. 借阅图书；
4. 保存到 JSON；
5. 新建一个图书馆对象；
6. 从 JSON 加载数据；
7. 展示加载后的图书。

````python
library = Library()

library.add_book(Book("Python基础", "张三"))
library.add_book(Book("数据分析入门", "李四"))
library.add_book(Book("机器学习实战", "王五"))

print("初始图书列表：")
library.show_books()

print("\n借阅一本书：")
library.borrow_book("Python基础")

print("\n借阅后的图书列表：")
library.show_books()

print("\n保存数据：")
library.save_to_file("library.json")

print("\n创建新的图书馆对象，并加载数据：")
new_library = Library()
new_library.load_from_file("library.json")

print("\n新图书馆中的图书列表：")
new_library.show_books()
````
