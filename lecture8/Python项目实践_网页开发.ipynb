# 二、第一个Flask程序

安装：

```bash

```

````python
from flask import Flask

app = Flask(__name__)

@app.route("/")
def home():
    return "Hello Python Web"

app.run()
````

代码解释：

`Flask(__name__)` 创建应用。

`@app.route("/")` 创建网页地址。

访问 `/` 会执行下面函数。

````python
from flask import Flask

app=Flask(__name__)

@app.route("/hello")
def hello():
    return "Hello Flask"

@app.route("/about")
def about():
    return "关于页面"

app.run()
````

# 三、返回HTML网页

真实网页需要HTML，而不是简单字符串。

````python
from flask import Flask
app=Flask(__name__)

@app.route("/")
def home():
    return """<h1>Python Web项目</h1><p>Flask开发成功</p>"""

app.run()
````

实际项目中HTML通常放在templates目录：

project
├ app.py
└ templates
  └ index.html

Python负责逻辑，HTML负责展示。

````python
from flask import Flask,render_template

app=Flask(__name__)

@app.route("/")
def home():
    return render_template("index.html")

app.run()
````

# 四、动态路由

网页经常需要根据用户输入返回不同内容。

例如：

/user/Tom

显示：你好Tom

````python
from flask import Flask
app=Flask(__name__)

@app.route("/user/<name>")
def user(name):
    return f"你好 {name}"

app.run()
````

# 五、综合案例：学生成绩查询网站

访问：

/student/Tom

返回学生成绩。

````python
from flask import Flask
app=Flask(__name__)

students={"Tom":95,"Alice":88,"Jack":76}

@app.route("/student/<name>")
def student(name):
    score=students.get(name)
    if score:
        return f"学生:{name}<br>成绩:{score}"
    return "学生不存在"

app.run()
````

# 练习2参考答案

````python
from flask import Flask
app=Flask(__name__)

products={"apple":5,"banana":3,"milk":8}

@app.route("/product/<name>")
def product(name):
    price=products.get(name)
    if price:
        return f"商品:{name}<br>价格:{price}元"
    return "商品不存在"

app.run()
````
