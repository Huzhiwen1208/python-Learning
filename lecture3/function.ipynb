```python
def add_all(*numbers):
    total = 0
    for number in numbers:
        total += number
    return total

print(add_all(1, 2, 3, 4, 5))
```

```python
def get_level(score: int) -> str:
    if score >= 90:
        return "优秀"
    elif score >= 80:
        return "良好"
    elif score >= 60:
        return "及格"
    else:
        return "不及格"


def print_student_result(name: str, score: int) -> None:
    level = get_level(score)
    print(f"{name} 的成绩是 {score}，等级是 {level}")

print_student_result("Tom", 95)
print_student_result("Alice", 82)
print_student_result("Jack", 55)
```

任务二：自学一下常用内置函数，不理解的下节课讲。

```python
scores = {
    "Tom": 95,
    "Alice": 88,
    "Jack": 76
}
def average_score(scores:float):
    total = sum(scores.values())
    count = len(scores)
    average = total / count
    return average

result = average_score(scores)
print(result)
```
