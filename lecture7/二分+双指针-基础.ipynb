```python
# 有一场景：我需要不断的从一个 list 中寻找到某个数，要求能快速找到.
num_list = [1, 5, 4, 3, 2, 6]

found = 2

# 暴力寻找，比较次数多，在极端场景，对于长度为 N 的列表，找最后一个元素，要找 N 次
for i, num in enumerate(num_list):
    if num == found:
        print(f'Found: {num} in {i}')

# 二分查找引入：现有一个天平和12个金币，但是有且只有一个假币，假币比较轻
# 你需要称多少次天平才可以找到这个假币
num_list.sort()
print(num_list)

def binary_search(nums: list[int], target: int) -> int:
    left = 0
    right = len(nums) - 1
    while left <= right:
        middle = (left + right) // 2
        if nums[middle] == target:
            return middle
        elif nums[middle] > target:
            right = middle - 1
        else:
            left = middle + 1
    return -1

print(binary_search(num_list, 2))
print(binary_search(num_list, 5))
print(binary_search(num_list, 7))
```

```python
# 双指针：https://leetcode.cn/problems/kLl5u1/
class Solution:
    def twoSum(self, numbers: List[int], target: int) -> List[int]:
        left  = 0
        right = len(numbers) - 1

        while left <= right:
            sum = numbers[left] + numbers[right]
            if sum == target:
                return [left, right]
            elif sum > target:
                right -= 1
            else:
                left  += 1
        
        return [-1, -1]
```
